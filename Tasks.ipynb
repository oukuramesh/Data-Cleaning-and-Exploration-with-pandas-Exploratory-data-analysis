```python
# find the departaments
df['Departament'].value_counts()
```

```python
# find the highest salary 
print("Highest_Salary:",df["Salary"].max())
```

```python
# # find the lowest salary 
print("Lowest_Salary:",df["Salary"].min())
```

```python
df.groupby("Departament")["Salary"].mean()
```

```python
df.groupby("Departament")["Salary"].max()
```

```python
df[df["City"] == "Hyderabad"]
```

```python
df.groupby("Departament")["Salary"].mean().plot(
    kind="bar"
)

plt.title("Average Salary by Departament")
plt.xlabel("Departament")
plt.ylabel("Average Salary")
plt.show()
```

```python
df.groupby("Departament")["Salary"].max().plot(
    kind="bar"
)

plt.title("Average Salary by Departament")
plt.xlabel("Departament")
plt.ylabel("Average Salary")
plt.show()
```

```python
df.groupby("Name")["Salary"].max().plot(
    kind="bar"
)

plt.title("Average Salary by Departament")
plt.xlabel("Name")
plt.ylabel("Average Salary")
plt.show()
```
